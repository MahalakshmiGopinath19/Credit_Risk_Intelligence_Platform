```sql
-- ============================================================
-- CREDIT RISK INTELLIGENCE PLATFORM
-- BRONZE LAYER - RAW DATA INGESTION
-- Source: Amazon S3
-- ============================================================

-- 1. Create Bronze schema
--CREATE SCHEMA IF NOT EXISTS credit_risk_bronze;


-- ============================================================
-- 2. APPLICATION(S)
-- ============================================================

CREATE OR REPLACE TABLE credit_risk_bronze.`application(s)`
AS
SELECT *
FROM read_files(
    's3://credit-risk-intelligence-platform/raw/application(s).csv',
    format => 'csv',
    header => true,
    inferSchema => true
);


-- ============================================================
-- 3. BUREAU
-- ============================================================

CREATE OR REPLACE TABLE credit_risk_bronze.bureau
AS
SELECT *
FROM read_files(
    's3://credit-risk-intelligence-platform/raw/bureau.csv',
    format => 'csv',
    header => true,
    inferSchema => true
);


-- ============================================================
-- 4. PREVIOUS APPLICATION
-- ============================================================

CREATE OR REPLACE TABLE credit_risk_bronze.previous_application
AS
SELECT *
FROM read_files(
    's3://credit-risk-intelligence-platform/raw/previous_application.csv',
    format => 'csv',
    header => true,
    inferSchema => true
);


-- ============================================================
-- 5. POS CASH BALANCE
-- ============================================================

CREATE OR REPLACE TABLE credit_risk_bronze.pos_cash_balance
AS
SELECT *
FROM read_files(
    's3://credit-risk-intelligence-platform/raw/POS_CASH_balance.csv',
    format => 'csv',
    header => true,
    inferSchema => true
);


-- ============================================================
-- 6. CREDIT CARD BALANCE
-- ============================================================

CREATE OR REPLACE TABLE credit_risk_bronze.credit_card_balance
AS
SELECT *
FROM read_files(
    's3://credit-risk-intelligence-platform/raw/credit_card_balance.csv',
    format => 'csv',
    header => true,
    inferSchema => true
);


-- ============================================================
-- 7. INSTALLMENTS PAYMENTS
-- ============================================================

CREATE OR REPLACE TABLE credit_risk_bronze.installments_payments
AS
SELECT *
FROM read_files(
    's3://credit-risk-intelligence-platform/raw/installments_payments.csv',
    format => 'csv',
    header => true,
    inferSchema => true
);


-- ============================================================
-- 8. BUREAU BALANCE
-- ============================================================

CREATE OR REPLACE TABLE credit_risk_bronze.bureau_balance
AS
SELECT *
FROM read_files(
    's3://credit-risk-intelligence-platform/raw/bureau_balance.csv',
    format => 'csv',
    header => true,
    inferSchema => true
);
```
